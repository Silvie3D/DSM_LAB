```python
import numpy as np
a = np.array([[1,2],[3,4]])
b = np.array([[5,6],[7,8]])

add = (a + b)
print("Addition: ", add,"\n")

diff = (a - b)
print("Difference: ", diff,"\n")

dot = (a * b)
print("Dot Product: ", dot,"\n")

trans_a = np.transpose(a)
trans_b = np.transpose(b)
print("Transpose: \n", trans_a,"\n\n", trans_b,"\n")

cross = np.matmul(a,b)
print("Cross Product: ", cross,"\n")
```

```python
import numpy as np
a = np.array([[1,2],[3,4]])
U, S, VT = np.linalg.svd(a)
n_comps = 2
a_recons = np.dot(U[:,:n_comps], np.dot(np.diag(S[:n_comps]),VT[:n_comps]))
print("Original Matrix: \n", a)
print("\nReconstrcted Matrix(with reduced dimensions):")
print(a_recons)
```
